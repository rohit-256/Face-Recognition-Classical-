```python
training_faces = faces[:,:,:3*160]
test_faces = faces[:,:,3*160:]
print(training_faces.shape)
print(test_faces.shape)
```

### Parametric Estimation

```python
train_flat = training_faces.reshape(-1, training_faces.shape[2])
print(train_flat.shape)
sample_mean = np.mean(train_flat, axis = 1, keepdims = True)
print(sample_mean.shape)
sample_covariance = (1/train_flat.shape[1]) * (train_flat - sample_mean)@(train_flat - sample_mean).T
sample_covariance.shape
```

```python
gaussian_vector_mv = np.random.multivariate_normal(sample_mean.flatten(), sample_covariance, size=1)
a = gaussian_vector_mv.reshape(24, 21)
plot = plt.imshow(a)
plt.show
```
